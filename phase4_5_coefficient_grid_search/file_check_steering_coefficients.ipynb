```python
import pandas as pd
import os
from pathlib import Path
import glob

# Set pandas display options
pd.set_option('display.max_colwidth', None)
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)
pd.set_option('display.width', None)

# Auto-discovery of Phase 4.5 steering coefficient data
datasets_dir = "../data/phase4_5/"
pattern = os.path.join(datasets_dir, "*.parquet")
matching_files = sorted(glob.glob(pattern))

if not matching_files:
    raise FileNotFoundError(f"No parquet files found in {datasets_dir}")

print(f"Found {len(matching_files)} files")
```

```python
# Load and display each dataset
for file_path in matching_files:
    df = pd.read_parquet(file_path)
    print(f"\n## {Path(file_path).name}: {len(df)} records")
    display(df)
```
